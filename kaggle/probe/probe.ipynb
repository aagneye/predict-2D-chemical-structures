```python
# Probe the scored-run environment: no internet, so every dependency must
# already be present or pre-attached as a Dataset.
import sys, platform, importlib, os
print('python', sys.version.split()[0], platform.machine())
for name in ['numpy','pandas','pyarrow','numba','sklearn','rdkit','torch','scipy']:
    try:
        m = importlib.import_module(name)
        print(f'{name:10s} {getattr(m, "__version__", "?")}')
    except Exception as e:
        print(f'{name:10s} MISSING ({type(e).__name__})')
print('--- cpu ---')
print('cpu count', os.cpu_count())
print('--- competition data ---')
for root, dirs, files in os.walk('/kaggle/input'):
    for f in files:
        p = os.path.join(root, f)
        print(f'{os.path.getsize(p)/1e6:10.1f} MB  {p}')
print('--- rdkit tautomer check ---')
try:
    from rdkit import Chem
    from rdkit.Chem.MolStandardize import rdMolStandardize
    e = rdMolStandardize.TautomerEnumerator()
    m = e.Canonicalize(Chem.MolFromSmiles('CC(=O)C'))
    print('inchikey14', Chem.MolToInchiKey(m)[:14])
except Exception as ex:
    print('rdkit tautomer FAILED:', ex)
```
